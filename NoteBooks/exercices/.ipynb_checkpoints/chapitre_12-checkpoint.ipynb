```python
# Chapitre 12:
# exercice 1:
try:
    x = 1 / 0
except Exception as e:
    print("Erreur capturée:", e)
```

```python
# exercice 2:
try:
    x = int("abc")
except ValueError:
    print("Conversion invalide")
except Exception as e:
    print("Autre erreur:", e)
```

```python
# exercice 3:
try:
    x = 10 / 2
except Exception as e:
    print(e)
else:
    print("Succès:", x)
```

```python
# exercice 4:
try:
    x = 5 / 1
except:
    print("Erreur")
finally:
    print("Bloc finally exécuté")
```

```python
# exercice 5:
class MyError(Exception):
    pass

def check(x):
    if x < 0:
        raise MyError("Nombre négatif interdit")
    return x

try:
    check(-1)
except MyError as e:
    print(e)
```

```python
# exercice 6:
def divide(a, b):
    if b == 0:
        raise ValueError("Division par zéro")
    return a / b

try:
    divide(4, 0)
except ValueError as e:
    print(e)
```

```python
# exercice 7:
with open("test.txt", "w") as f:
    f.write("hello world")

with open("test.txt", "r") as f:
    print(f.read())
```

```python
# exercice 8:
try:
    x = int("abc")
except (ValueError, TypeError) as e:
    print("Erreur multiple:", e)
```

```python
# exercice 9:
def check_positive(x):
    assert x > 0, "Doit être positif"
    return x

try:
    check_positive(-5)
except AssertionError as e:
    print(e)
```

```python
# exercice 10:
def debug_example(x):
    print("Entrée:", x)
    result = x * 2
    print("Résultat:", result)
    return result

debug_example(5)
```
