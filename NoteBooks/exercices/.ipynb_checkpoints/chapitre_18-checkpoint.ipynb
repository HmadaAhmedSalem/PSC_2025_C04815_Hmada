```python
# Chapitre 18:
# exercice 1:
from multiprocessing import Process

def task():
    print("Process running")

p = Process(target=task)
p.start()
p.join()
```

```python
# exercice 2:
from multiprocessing import Process

def worker(i):
    print(f"Worker {i}")

processes = []
for i in range(3):
    p = Process(target=worker, args=(i,))
    processes.append(p)
    p.start()

for p in processes:
    p.join()
```

```python
# exercice 3:
from multiprocessing import Pool

def square(x):
    return x*x

with Pool(4) as p:
    print(p.map(square, [1,2,3,4]))
```

```python
# exercice 4:
import os
print("CPU cores:", os.cpu_count())
```

```python
# exercice 5:
from multiprocessing import Pool

with Pool(2) as p:
    print(p.map(lambda x: x+1, [1,2,3]))
```

```python
# exercice 6:
from multiprocessing import Pool

def add_one(x):
    return x+1

with Pool(2) as p:
    result = p.apply_async(add_one, (5,))
    print(result.get())
```

```python
# exercice 7:
import time

def slow_square(x):
    time.sleep(1)
    return x*x

start = time.time()
for i in [1,2,3]:
    slow_square(i)
print("Sequential:", time.time()-start)

from multiprocessing import Pool
start = time.time()
with Pool(3) as p:
    p.map(slow_square, [1,2,3])
print("Parallel:", time.time()-start)
```

```python
# exercice 9:
from multiprocessing import Process, Queue

def worker(q):
    q.put("data from process")

q = Queue()
p = Process(target=worker, args=(q,))
p.start()
print(q.get())
p.join()
```
