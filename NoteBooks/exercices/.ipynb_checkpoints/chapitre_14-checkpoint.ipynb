```python
# Chapitre 14:
# exercice 1:
with open("file.txt", "w") as f:
    f.write("Hello World")
```

```python
# exercice 2:
with open("file.txt", "r") as f:
    print(f.read())
```

```python
# exercice 3:
with open("file.txt", "a") as f:
    f.write("\nNew line added")
```

```python
# exercice 4:
with open("file.txt", "r") as f:
    for line in f:
        print(line.strip())
```

```python
# exercice 5:
data = ["a", "b", "c"]
with open("list.txt", "w") as f:
    for item in data:
        f.write(item + "\n")
```

```python
# exercice 6:
with open("list.txt", "r") as f:
    lines = f.readlines()
print(lines)
```

```python
# exercice 7:
data = bytes([120, 3, 255])
with open("binary.bin", "wb") as f:
    f.write(data)

with open("binary.bin", "rb") as f:
    print(f.read())
```

```python
# exercice 8:
import json

obj = {"name": "Hmada", "age": 35}
with open("data.json", "w") as f:
    json.dump(obj, f)

with open("data.json", "r") as f:
    print(json.load(f))
```

```python
# exercice 9:
import os

print(os.path.exists("file.txt"))
```
