```python
# Chapitre 11:
# exercice 1:
import matplotlib.pyplot as plt

x = [1,2,3]
y = [4,5,6]

plt.plot(x, y)
plt.show()
```

```python
# exercice 2:
plt.plot(x, y)
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("Simple Plot")
plt.show()
```

```python
# exercice 3:
y2 = [6,5,4]
plt.plot(x, y)
plt.plot(x, y2)
plt.show()
```

```python
# exercice 4:
plt.scatter(x, y)
plt.show()
```

```python
# exercice 5:
plt.bar(x, y)
plt.show()
```

```python
# exercice 6:
from matplotlib.widgets import Slider

fig, ax = plt.subplots()
plt.subplots_adjust(bottom=0.25)

t = [0,1,2,3]
s = [1,2,3,4]
l, = plt.plot(t, s)

ax_slider = plt.axes([0.2, 0.1, 0.65, 0.03])
slider = Slider(ax_slider, 'Scale', 1, 5, valinit=1)

def update(val):
    scale = slider.val
    l.set_ydata([i*scale for i in s])
    fig.canvas.draw_idle()

slider.on_changed(update)
plt.show()
```

```python
# exercice 7:
from matplotlib.widgets import Button

fig, ax = plt.subplots()
l, = plt.plot([1,2,3], [4,5,6])

def change(event):
    l.set_ydata([6,5,4])
    plt.draw()

ax_button = plt.axes([0.8, 0.05, 0.1, 0.075])
btn = Button(ax_button, 'Change')
btn.on_clicked(change)

plt.show()
```
