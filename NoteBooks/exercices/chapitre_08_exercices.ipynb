```python
# Chapitre 8:
# exercice 1:
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p = Person("Hmada", 35)
print(p.name, p.age)
```

```python
# exercice 2:
class Rectangle:
    def __init__(self, w, h):
        self.w = w
        self.h = h

    def area(self):
        return self.w * self.h

r = Rectangle(4,5)
print(r.area())
```

```python
# exercice 3:
class Circle:
    def __init__(self, r):
        self.r = r

    def area(self):
        return 3.14 * self.r**2

c = Circle(3)
print(c.area())
```

```python
# exercice 4:
class Counter:
    count = 0

    def __init__(self):
        Counter.count += 1

a = Counter()
b = Counter()
print(Counter.count)
```

```python
# exercice 5:
class Animal:
    def speak(self):
        return "Some sound"

class Dog(Animal):
    def speak(self):
        return "Bark"

d = Dog()
print(d.speak())
```

```python
# exercice 6:
class BankAccount:
    def __init__(self, balance):
        self._balance = balance

    def deposit(self, amount):
        self._balance += amount

    def withdraw(self, amount):
        self._balance -= amount

    def get_balance(self):
        return self._balance

acc = BankAccount(100)
acc.deposit(50)
acc.withdraw(30)
print(acc.get_balance())
```

```python
# exercice 7:
class Temperature:
    def __init__(self, c):
        self._c = c

    @property
    def celsius(self):
        return self._c

    @celsius.setter
    def celsius(self, value):
        self._c = value

t = Temperature(25)
print(t.celsius)
t.celsius = 30
print(t.celsius)
```
