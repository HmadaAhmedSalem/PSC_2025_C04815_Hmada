```python
# exercice 2:
def greet(name, age):
    return f"Name: {name}, Age: {age}"

# Positionnel
print(greet("Hmada", 35))

# Nommé
print(greet(age=35, name="Hmada"))
```

```python
# exercice 3:
def greet_default(name="Inconnu"):
    return f"Hello {name}"

print(greet_default())
print(greet_default("Hmada"))
```

```python
# exercice 4:
def sum_all(*args):
    return sum(args)

print(sum_all(1, 2, 3, 4))
```

```python
# exercice 5:
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n-1)

print(factorial(5))
```

```python
# exercice 6:
square = lambda x: x**2

print(square(4))
```
