```python
# Chapitre 16:
# exercice 1:
from sympy import symbols

x = symbols('x')
expr = x**2 + 2*x + 1
print(expr)
```

```python
# exercice 2:
from sympy import expand

expr = (x + 1)**2
print(expand(expr))
```

```python
# exercice 3:
from sympy import factor

expr = x**2 + 2*x + 1
print(factor(expr))
```

```python
# exercice 4:
from sympy import solve

solutions = solve(x**2 - 4, x)
print(solutions)
```

```python
# exercice 5:
from sympy import diff

expr = x**3
print(diff(expr, x))
```

```python
# exercice 6:
from sympy import integrate

expr = x**2
print(integrate(expr, x))
```

```python
# exercice 7:
expr = x**2 + 1
print(expr.subs(x, 2))
```

```python
# exercice 8:
from sympy import lambdify

f = lambdify(x, x**2 + 1)
print(f(3))
```

```python
# exercice 9:
y = symbols('y')
expr = x + y
print(expr.subs({x:1, y:2}))
```

```python
# exercice 10:
from sympy import simplify

expr = (x**2 + 2*x + 1)/(x+1)
print(simplify(expr))
```
