```python
# Chapitre 13:
# exercice 1:
x = 10

def show():
    print("Local access to global x:", x)

show()
```

```python
# exercice 2:
x = 10

def func():
    x = 5
    print("Local x:", x)

func()
print("Global x:", x)
```

```python
# exercice 3:
x = 10

def modify():
    global x
    x = 20

modify()
print(x)
```

```python
# exercice 4:
def outer():
    x = "outer"

    def inner():
        print(x)

    inner()

outer()
```

```python
# exercice 5:
def outer():
    x = 10
    def inner():
        nonlocal x
        x = 20
    inner()
    print(x)

outer()
```
